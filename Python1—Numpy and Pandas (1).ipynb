### Data analysis using NumPy and Pandas ###

```python
# Numpy Array Operations:
# 1. Create a 1D NumPy Array:
import numpy as np
import pandas as pd
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
# 2. Inspection and Properties:
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
# 3. Array Operations:
temperatures_f1 = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", temperatures_f1)
print("Maximum temperature :", temperatures_w1.max())
print("Minimum temperature :", temperatures_w1.min())
print("Mean temperature :", temperatures_w1.mean())
```

```python
# 4. Array Slicing and Indexing:

print("First three days:", temperatures_w1[:3])
print("Weekend:", temperatures_w1[-2:])
print("Middle three days:", temperatures_w1[2:5])
```

```python
# 5. Create a 2D Array:
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])

print(temperatures)
```

```python
# 6. Inspect and Slice the 2D Array:

print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)

print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])
print("Week 1 weekend:", temperatures[0, -2:])
print("Week 2 weekend:", temperatures[1, -2:])
```

```python
# Pandas Series:
# 1. Creating Pandas Series:
# Create a Pandas Series named marks with the following data:
# Marks: 95, 92, 89, 85, 80
# Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

marks = pd.Series([95, 92, 89, 85, 80],index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"],name="marks")
print(marks)
```

```python
# 2. Indexing and Slicing:
# Use the integer index position to access the mark of the 1st rank student.
print(marks.iloc[0])

# Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index
#labels.
print(marks.loc[["Rank1", "Rank2", "Rank3"]])

# Use the iloc accessor to retrieve the mark of the 3rd-ranked student.
print(marks.iloc[2])

# Apply a boolean mask to filter the Series and retrieve the ranks where the marks are
#greater than 90.
print(marks[marks > 90])
```

```python
# 3. Manipulating Series:
# Modify the mark of the student with the 1st rank to 100.
marks.loc["Rank1"] = 100


# Remove the entry corresponding to the last rank from the Series.
marks = marks.drop("Rank5")


# Compute the CGPA by dividing each mark in the Series by 10.
cgpa = marks / 10
print(marks)
print(cgpa)
```

```python
# Pandas DataFrame:
# Creating Pandas DataFrame:

transactions = pd.DataFrame({"TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
                             "ProductCategory": ["Electronics","Clothing","Electronics","Furniture","Clothing","Electronics",
                                                 "Furniture","Clothing","Furniture","Electronics"],
                             "Region": ["North","South","North","East","West","North","East","West","South","North"],
                             "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]})

transactions
```

```python
# 2. Data Exploration:

print("First 5 rows:")
display(transactions.head())

print("Last 5 rows:")
display(transactions.tail())

print("Shape:", transactions.shape)
print("Column names:", transactions.columns.tolist())

print("Data types:")
print(transactions.dtypes)

print("Basic information:")
transactions.info()
```

```python
# 3. Manipulating the DataFrame:
# Modify Amount for TransactionID 102 to 165
transactions.loc[
    transactions['TransactionID'] == 102,
    'Amount'
] = 165

print("\nAfter changing TransactionID 102 Amount to 165:")
print(transactions)


# Add Discount column = 10% of Amount
transactions['Discount'] = transactions['Amount'] * 0.10

print("\nAfter adding Discount column:")
print(transactions)


# Remove TransactionID 109
transactions = transactions[
    transactions['TransactionID'] != 109
].reset_index(drop=True)

print("\nAfter removing TransactionID 109:")
print(transactions)


# Delete Discount column
transactions = transactions.drop(columns=['Discount'])

print("\nAfter deleting Discount column:")
print(transactions)
```
